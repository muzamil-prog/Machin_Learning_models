```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('heart.csv')
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
df.head()
```

```python
dff['Sex']=le.fit_transform(df['Sex'])
```

```python
dff.head()
```

```python
dff['ChestPainType']=le.fit_transform(dff['ChestPainType'])
dff['RestingECG']=le.fit_transform(dff['RestingECG'])

```

```python
dff.head()

```

```python
dff.head()
```

```python
dff['ST_Slope']=le.fit_transform(dff['St_Slope'])
```

```python
dff.head()
```

```python
x=dff.drop("HeartDisease",axis=1)
y=dff['HeartDisease']
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
```

```python
model.fit(x_train, y_train)
```

```python
y_predict=model.predict(x_test)
```

```python
y_predict
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
accuracy_score(y_test, y_predict)
```

```python
confusion_matrix(y_test, y_predict)
```

```python
print(classification_report(y_test, y_predict))
```

# KNN(Nearest Neighber)

## Feature Scaling


```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.fit_transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_model=KNeighborsClassifier(n_neighbors=5)
```

```python
knn_model.fit(x_train_scaled, y_train)
```

```python
k_y_predict=knn_model.predict(x_test_scaled)
```

```python
k_y_predict
```

```python
accuracy_score(k_y_predict, y_test)
```

```python
confusion_matrix(k_y_predict, y_test)
```

```python
print(classification_report(k_y_predict, y_test))
```
