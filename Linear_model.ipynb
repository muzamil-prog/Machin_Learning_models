```python
x=df.drop(columns=['charges'])
```

```python
y=df['charges']
```

```python
x
```

```python
y
```

```python
one_hot_encode=pd.get_dummies(x, columns=['sex','smoker', 'region'], drop_first=True)
```

```python
x
```

```python
one_hot_encode
```

```python
one_hot_encode=one_hot_encode.astype(int)
```

```python
one_hot_encode
```

```python
from sklearn.preprocessing import LabelEncoder
encode=LabelEncoder()
columns=['sex', 'smoker','region']
x_label=x
for i in columns:
    x_label[i]=encode.fit_transform(x_label[i])

```

```python
x_label
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
x_train, x_test, y_train, y_test=train_test_split(one_hot_encode, y ,test_size=0.2, random_state=44)
```

```python
model=LinearRegression()
model.fit(x_train, y_train)
```

```python
y_predict=model.predict(x_test)
```

```python
y_predict
```

```python
y_test
```

```python
r2=r2_score(y_test, y_predict)
```

```python
r2
```
